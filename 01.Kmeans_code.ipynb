```python
from sklearn.cluster import KMeans
```

```python
distortions = []
K_range = range(1, 11)

for i in K_range:
    kmeans = KMeans(n_clusters=i)
    kmeans.fit(d)
    distortions.append(kmeans.inertia_)
```

```python
plt.figure(figsize=(8, 4))
plt.plot(K_range, distortions, marker='o')
plt.title('Elbow Method for Optimal k')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Distortion')
plt.show()
```

```python
kn = 3

kmeans = KMeans(n_clusters=kn, random_state=10)
kmeans.fit(d)

score_means = kmeans.cluster_centers_.mean(axis=1)
cluster_order = np.argsort(score_means)
label_map = {cluster_order[0]: 'Mild',
             cluster_order[1]: 'Moderate',
             cluster_order[2]: 'Severe'}
labels = pd.Series(kmeans.labels_).map(label_map)
```

```python
labels = pd.Series(kmeans.labels_).map(label_map)
```

```python
merged_df['Cluster'] = labels
```

```python
cluster_centers = kmeans.cluster_centers_[cluster_order]
```

```python
cluster_centers
```

```python
x_values = merged_df.iloc[:,1:-1].columns
```

```python
plt.figure(figsize=(10, 8))
    
plt.plot(x_values, cluster_centers[0], label='Mild', marker='o',color='blue')
plt.plot(x_values, cluster_centers[1], label='Moderate', marker='o',color='orange')
plt.plot(x_values, cluster_centers[2], label='Severe', marker='o',color='red')



plt.xlabel('Feature')
plt.ylabel('Cluster Center Value')
plt.title('Cluster Centers for Each Feature')
plt.legend(loc=4)
plt.xticks(rotation=90)
plt.show()
```
